```python
df2_sorted = df2.sort_values(by='연월', ascending=True).reset_index(drop=True)

df2_sorted
```

```python
year_month = df2.groupby(['연도', '월'])['물동량'].sum().reset_index()
year_month
```

```python
year_avg = year_month.groupby('연도')['물동량'].transform('mean')
```

```python
year_month['평균대비(%)'] = (
    year_month['물동량'] / year_avg * 100
)

year_month
```

```python
monthly_pattern = year_month.groupby('월')['평균대비(%)'].mean()

monthly_pattern
```

```python
monthly = (
df2.groupby(['연도', '월'])['물동량'].sum().reset_index()
)

monthly_avg = (monthly.groupby('월')['물동량'].mean().reset_index()
)

monthly_avg['증감률(%)'] = monthly_avg['물동량'].pct_change() * 100

monthly_avg
```

```python
avg = monthly_avg['물동량'].mean()

monthly_avg['평균차이'] = monthly_avg['물동량'] - avg
```

```python
monthly_avg['차이율(%)'] = (monthly_avg['평균차이'] / avg * 100).round(2)
monthly_avg
```

- 월별 물동량은 전체 평균을 기준으로 일정한 차이를 보이고 있음
- 특히 2월은 전체 월평균보다 -8.37%로 가장 낮고, 9월 또한 -3.06%로 낮음
- 반면에 3월은 3.37%로 가장 높게 나타남

# 환적 구조 파악

```python
df3 = pd.read_csv(
    rr_CSV_PATH, encoding='cp949')
df3
```

- 부산항의 전체 물동량이 증가하는 가운데, 수출입물동량보다 환적물동량이 빠르게 증가한다는 것을 알 수 있음.

```python
print(df3.columns.tolist())
```

```python
result = df3[[
    '연도',
    ' 컨물동량',
    '수출입소계',
    '환적소계',
    '환적점유율'
]]

result
```

```python
df3['전체증가량'] = df3[' 컨물동량'].diff()
df3['환적증가량'] = df3['환적소계'].diff()

df3['환적기여율'] = (df3['환적증가량'] / df3['전체증가량'] * 100).round(2)

df3[['연도', ' 컨물동량', '전체증가량', '환적증가량', '환적점유율', '환적기여율']]
```
