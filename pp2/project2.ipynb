```python
df['전체증가량'] = df[' 컨물동량'].diff()

df['환적증가량'] = df['환적소계'].diff()

df['환적기여율'] = (df['환적증가량'] / df['전체증가량']*100).round(2)

df[['연도', ' 컨물동량', '전체증가량', '환적증가량', '환적점유율', '환적기여율']]
```

## 결론
부산항은 전체 컨테이너 물동량에서 환적의 비중이 지속적으로 확대되고 있으며, 최근 물동량 성장 역시 환적이 주도하는 구조가 뚜렷하게 나타나고 있다.
